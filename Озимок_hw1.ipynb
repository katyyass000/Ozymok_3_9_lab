```python
import numpy as np
```

```python
# 1. Створіть одновимірний масив (вектор) з першими 20-ма натуральними числами та виведіть його значення.
vector_nat = np.arange(1, 21)
print(vector_nat)
```

```python
# 2. Створіть двовимірний масив (матрицю) розміром 4x4, заповніть його одиницями та виведіть його значення.
matrix_ones = np.ones((4, 4))
print(matrix_ones)
```

```python
# 3. Створіть масив розміром 5x5, заповніть його випадковими дійсними числами в діапазоні від -1 до 1 та виведіть його значення.
matrix_rand_5x5 = np.random.uniform(-1, 1, (5, 5))
print(matrix_rand_5x5)
```

```python
# 4. Створіть масив розміром 3x3, заповніть його випадковими дійсними числами в діапазоні від 0 до 1 та виведіть його значення.
matrix_rand_3x3 = np.random.rand(3, 3)
print(matrix_rand_3x3)
```

```python
# 5. Створіть два одновимірних масиви розміром 6, заповніть їх випадковими цілими числами
# в діапазоні від 1 до 20 та виконайте на них поелементні операції додавання, віднімання та множення.
arr1 = np.random.randint(1, 21, 6)
arr2 = np.random.randint(1, 21, 6)

print(f"Масив 1: {arr1}")
print(f"Масив 2: {arr2}")
print(f"Додавання: {arr1 + arr2}")
print(f"Віднімання: {arr1 - arr2}")
print(f"Множення:  {arr1 * arr2}")
```

```python
# 6. Створіть дві матриці розміром 3x2 та 2x4, заповніть їх випадковими цілими числами
# в діапазоні від 1 до 5 та перемножте їх між собою.
mat_a = np.random.randint(1, 6, (3, 2))
mat_b = np.random.randint(1, 6, (2, 4))
res_dot = np.dot(mat_a, mat_b)

print("Матриця 3x2:\n", mat_a)
print("Матриця 2x4:\n", mat_b)
print("Результат множення:\n", res_dot)
```

```python
# 7. Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами в діапазоні від 0 до 1 та транспонуйте її.
mat_trans = np.random.rand(3, 3)
print("Оригінальна матриця:\n", mat_trans)
print("Транспонована матриця:\n", mat_trans.T)
```

```python
# 8. Створіть дві матриці розміром 3x3, заповніть їх випадковими цілими числами
# в діапазоні від 1 до 10 та знайдіть їхню різницю.
mat_m1 = np.random.randint(1, 11, (3, 3))
mat_m2 = np.random.randint(1, 11, (3, 3))

print("Матриця 1:\n", mat_m1)
print("Матриця 2:\n", mat_m2)
print("Різниця:\n", mat_m1 - mat_m2)
```

```python
# 9. Створіть дві матриці розміром 3x3, заповніть їх випадковими цілими числами
# в діапазоні від -5 до 5 та знайдіть їх добуток.
mat_p1 = np.random.randint(-5, 6, (3, 3))
mat_p2 = np.random.randint(-5, 6, (3, 3))

print("Матриця 1:\n", mat_p1)
print("Матриця 2:\n", mat_p2)
print("Добуток матриць (dot product):\n", np.dot(mat_p1, mat_p2))
```

```python
# 10. Створіть матрицю розміром 3x3, заповніть її випадковими дійсними числами
# в діапазоні від -1 до 1 та знайдіть вектор-стовпчик, що містить суму елементів кожного рядка матриці.
mat_sum = np.random.uniform(-1, 1, (3, 3))
row_sums = np.sum(mat_sum, axis=1).reshape(-1, 1)

print("Матриця:\n", mat_sum)
print("Вектор-стовпчик сум рядків:\n", row_sums)
```
